```python
def normalize_speaker(s):
    s = s.strip().lower()
    if s in ["agent", "support", "rep"]:
        return "agent"
    if s in ["customer", "client", "user"]:
        return "customer"
    return "unknown"

def normalize_outcome(o):

    o = o.strip().lower()

    if "escalat" in o:
        return "escalation"
    if "complaint" in o:
        return "complaint"
    if "refund" in o:
        return "refund"

    return "other"

```

```python
normalized_conversations = []

for conv in raw_data['transcripts']:
    call_id = conv["transcript_id"]
    outcome = normalize_outcome(conv["intent"])

    turns = []
    for i, t in enumerate(conv["conversation"], start=1):
        turns.append({
            "turn_id": i,
            "speaker": normalize_speaker(t["speaker"]),
            "text": t["text"].strip()
        })

    normalized_conversations.append({
        "call_id": call_id,
        "outcome": outcome,
        "turns": turns
    })

len(normalized_conversations)
```

```python
class Evidence:
    def __init__(self, call_id, turn_id, speaker, text, factor):
        self.call_id = call_id
        self.turn_id = turn_id
        self.speaker = speaker
        self.text = text
        self.factor = factor

class CausalExplanation:
    def __init__(self, call_ids, factors, evidence, explanation):
        self.call_ids = call_ids
        self.factors = factors
        self.evidence = evidence
        self.explanation = explanation
```

```python
CAUSAL_RULES = {
    "customer_frustration": [
        "already", "still pending", "no response", "not resolved",
        "this is unacceptable", "why is this taking", "again",
        "many times", "multiple times", "nobody helped",
        "i want to speak", "supervisor", "manager"
    ],

    "agent_non_resolution": [
        "please wait", "we are checking", "i will check",
        "i will forward", "i will escalate", "we are working on it",
        "kindly wait", "request raised", "ticket raised"
    ]
}
```

```python
def analyze_conversation(conv):
    factors_found = set()
    evidence = []

    for turn in conv["turns"]:
        text = turn["text"].lower()
        assigned_factor = None

        if turn["speaker"] == "agent":
            # Prioritize agent_non_resolution for agent turns
            for kw in CAUSAL_RULES["agent_non_resolution"]:
                if kw in text:
                    assigned_factor = "agent_non_resolution"
                    break
            # If no agent_non_resolution keyword found, still check customer_frustration (less likely for agent's own turn, but possible if agent repeats customer's frustration)
            if not assigned_factor:
                for kw in CAUSAL_RULES["customer_frustration"]:
                    if kw in text:
                        assigned_factor = "customer_frustration"
                        break
        elif turn["speaker"] == "customer":
            # Prioritize customer_frustration for customer turns
            for kw in CAUSAL_RULES["customer_frustration"]:
                if kw in text:
                    assigned_factor = "customer_frustration"
                    break
            # If no customer_frustration keyword found, check agent_non_resolution (less likely for customer's own turn)
            if not assigned_factor:
                for kw in CAUSAL_RULES["agent_non_resolution"]:
                    if kw in text:
                        assigned_factor = "agent_non_resolution"
                        break

        # Fallback for other speakers or if no speaker-specific factor assigned yet
        if not assigned_factor:
            for factor_key, keywords in CAUSAL_RULES.items():
                for kw in keywords:
                    if kw in text:
                        assigned_factor = factor_key
                        break
                if assigned_factor:
                    break

        if assigned_factor:
            factors_found.add(assigned_factor)
            evidence.append(
                Evidence(
                    call_id=conv["call_id"],
                    turn_id=turn["turn_id"],
                    speaker=turn["speaker"],
                    text=turn["text"],
                    factor=assigned_factor
                )
            )

    return factors_found, evidence
```

```python
all_factors = set()
all_evidence = []
call_ids = []

for conv in escalation_calls:
    factors, evidence = analyze_conversation(conv)
    if len(factors) >= 1 and len(evidence) >= 2:
       call_ids.append(conv["call_id"])
       all_factors |= factors
       all_evidence.extend(evidence)
```

```python
def build_causal_explanation(call_ids, factors, evidence):
    factor_to_evidence = defaultdict(list)
    for e in evidence:
        factor_to_evidence[e.factor].append(e)

    explanation_parts = []

    if "customer_frustration" in factor_to_evidence:
        examples = factor_to_evidence["customer_frustration"][:2]
        quotes = "; ".join([f'"{e.text}"' for e in examples])
        explanation_parts.append(
            f"Customers showed repeated frustration, for example {quotes}."
        )

    if "agent_non_resolution" in factor_to_evidence:
        examples = factor_to_evidence["agent_non_resolution"][:2]
        quotes = "; ".join([f'"{e.text}"' for e in examples])
        explanation_parts.append(
            f"Agents responded with non-resolving actions such as {quotes}."
        )

    if (
        "customer_frustration" in factor_to_evidence and
        "agent_non_resolution" in factor_to_evidence
    ):
        explanation_parts.append(
            "The immediate sequence of customer frustration followed by non-resolving "
            "agent responses caused the interaction to escalate."
        )

    return " ".join(explanation_parts)
```

```python
explanation_text = build_causal_explanation(
    call_ids=call_ids,
    factors=all_factors,
    evidence=all_evidence
)

task1_output = CausalExplanation(
    call_ids=call_ids,
    factors=list(all_factors),
    evidence=all_evidence,
    explanation=explanation_text
)

# -------- DISPLAY OUTPUT -------- #

print("===== TASK 1: CAUSAL EXPLANATION =====\n")

print("CALL IDS:")
print(task1_output.call_ids)

print("\nIDENTIFIED FACTORS:")
for f in task1_output.factors:
    print("-", f)

print("\nSUPPORTING EVIDENCE (sample):")
for e in task1_output.evidence[:6]:
    print(
        f"[Call {e.call_id} | Turn {e.turn_id} | {e.speaker}] "
        f"{e.text}  --> factor: {e.factor}"
    )

print("\nCAUSAL EXPLANATION:")
print(task1_output.explanation)
```

```python
def handle_followup(query, context):
    q = query.lower()

    if "why" in q or "escalat" in q:
        return {
            "type": "explanation",
            "content": context.explanation
        }

    if "factor" in q:
        return {
            "type": "factors",
            "content": context.factors
        }

    if "turn" in q or "evidence" in q:
        return {
            "type": "evidence",
            "content": [
                {
                    "call_id": e.call_id,
                    "turn_id": e.turn_id,
                    "speaker": e.speaker,
                    "text": e.text,
                    "factor": e.factor
                } for e in context.evidence
            ]
        }

    return {
        "type": "unknown",
        "content": "Query not supported by current context"
    }
```

## Identify Coherence Gaps

### Subtask:
Develop a function to identify conversational coherence gaps in escalation calls, where customer frustration is immediately followed by an agent's non-resolution response.


**Reasoning**:
The subtask requires defining a function `identify_coherence_gaps` to find instances where customer frustration is immediately followed by an agent's non-resolution response in escalation calls. This function will iterate through conversations and their turns, applying the `CAUSAL_RULES` to identify these patterns and store the relevant turns.



```python
def identify_coherence_gaps(escalation_calls):
    coherence_gaps = []

    for conv in escalation_calls:
        for i in range(len(conv['turns']) - 1):
            customer_turn = conv['turns'][i]
            agent_next_turn = conv['turns'][i+1]

            # Check for customer frustration in the current turn
            if customer_turn['speaker'] == 'customer':
                customer_frustration_keywords = CAUSAL_RULES.get("customer_frustration", [])
                is_customer_frustrated = False
                for kw in customer_frustration_keywords:
                    if kw in customer_turn['text'].lower():
                        is_customer_frustrated = True
                        break



                if is_customer_frustrated:
                    # Check for agent non-resolution in the immediate next turn
                    if agent_next_turn['speaker'] == 'agent':
                        agent_non_resolution_keywords = CAUSAL_RULES.get("agent_non_resolution", [])
                        is_agent_non_resolving = False
                        for kw in agent_non_resolution_keywords:
                            if kw in agent_next_turn['text'].lower():
                                is_agent_non_resolving = True
                                break

                        if is_agent_non_resolving:
                            coherence_gaps.append({
                                'call_id': conv['call_id'],
                                'customer_turn': customer_turn,
                                'agent_turn': agent_next_turn
                            })
    return coherence_gaps

print("Defined function 'identify_coherence_gaps'.")
```

```python
coherence_gaps = identify_coherence_gaps(escalation_calls)

print(f"Total coherence gaps identified: {len(coherence_gaps)}")

if coherence_gaps:
    print("\nExample of a coherence gap:")
    example_gap = coherence_gaps[0]
    print(f"Call ID: {example_gap['call_id']}")
    print(f"  Customer Turn {example_gap['customer_turn']['turn_id']} ({example_gap['customer_turn']['speaker']}): {example_gap['customer_turn']['text']}")
    print(f"  Agent Turn {example_gap['agent_turn']['turn_id']} ({example_gap['agent_turn']['speaker']}): {example_gap['agent_turn']['text']}")
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

embedder = SentenceTransformer("all-MiniLM-L6-v2")

def generate_relevant_answer(query, context):
    q = query.lower()

    if "why" in q:
        return context.explanation

    if "factor" in q:
        return "The main factors were: " + ", ".join(context.factors)

    if "evidence" in q or "turn" in q:
        return " ".join([e.text for e in context.evidence[:3]])

    return context.explanation
```

```python
def improved_faithfulness(explanation, evidence, threshold=0.5):
    sentences = [s.strip() for s in explanation.split(".") if s.strip()]
    evidence_texts = [e.text for e in evidence]

    if not sentences or not evidence_texts:
        return 0.0

    grounded = 0

    for sent in sentences:
        sent_emb = embedder.encode([sent])
        max_sim = 0

        for ev in evidence_texts:
            ev_emb = embedder.encode([ev])
            sim = cosine_similarity(sent_emb, ev_emb)[0][0]
            max_sim = max(max_sim, sim)

        if max_sim >= threshold:
            grounded += 1

    return grounded / len(sentences)
```

```python
def strict_faithfulness(explanation, evidence, threshold=0.55):
    """
    Sentence-level grounding check.
    """
    evidence_text = " ".join([e.text for e in evidence])
    ev_emb = embedder.encode([evidence_text])

    sentences = [s.strip() for s in explanation.split(".") if s.strip()]
    grounded = 0

    for s in sentences:
        s_emb = embedder.encode([s])
        sim = cosine_similarity(s_emb, ev_emb)[0][0]
        if sim >= threshold:
            grounded += 1

    return grounded / max(len(sentences), 1)
```

```python
def evidence_composed_answer(query, context, max_evidence=3):
    """
    Generates a faithfulness-safe answer by composing extracted evidence.
    NO abstraction. NO inference. NO new claims.
    """

    q = query.lower()
    selected = []

    if "why" in q or "escalate" in q:
        selected = context.evidence[:max_evidence]

    elif "agent" in q:
        selected = [e for e in context.evidence if e.speaker == "agent"][:max_evidence]

    elif "customer" in q:
        selected = [e for e in context.evidence if e.speaker == "customer"][:max_evidence]

    else:
        selected = context.evidence[:max_evidence]

    answer = " ".join([e.text for e in selected])
    return answer
```

```python
def intent_aligned_answer(query, context, max_evidence=3):
    """
    Generates a relevance-optimized AND faithful answer.
    One causal sentence + evidence stitching.
    """

    q = query.lower()

    # 1️⃣ Minimal causal anchor (NO new info)
    if "why" in q or "escalate" in q:
        anchor = (
            "These calls escalated due to repeated customer frustration "
            "followed by agent responses that did not resolve the issue."
        )
    else:
        anchor = "The following evidence addresses the query."

    # 2️⃣ Evidence (faithfulness preserved)
    evidence_snippets = " ".join(
        [e.text for e in context.evidence[:max_evidence]]
    )

    return anchor + " " + evidence_snippets
```
